# Slovníky v Pythone

## 1. Čo sú slovníky?

Slovník je užitočná dátová štruktúra, ktorá uchováva páry **kľúč – hodnota**.

- **Kľúč** je obvykle textový reťazec (môže byť aj iný dátový typ)
- **Hodnota** môže byť akýkoľvek dátový typ (číslo, reťazec, zoznam, dokonca aj ďalší slovník)


## 2. Definícia slovníka

**Syntax:**
- Slovník začína a končí **zloženou zátvorkou** `{ }`
- Kľúč oddeľujeme od hodnoty **dvojbodkou** `:`
- Jednotlivé položky oddeľujeme **čiarkou** `,`

Pravidlá:
- Každý kľúč musí byť jedinečný
- Hodnoty sa môžu opakovať
- Hodnota môže byť zoznam (napr. zoznam postáv)


```python
# Príklad: informácie o knihe
kniha = {
    "nazov": "Ilias",
    "pocet_stran": 200,
    "postavy": ["Odysseus", "Patroklos", "Ajax", "Hektor"]
}
print(kniha)
```

```python
# Alternatívne: pomocou dict() s keyword argumentmi
kniha = dict(nazov="Ilias",
             pocet_stran=200,
             postavy=["Odysseus", "Patroklos", "Ajax", "Hektor"])
print(kniha)
```

## 4. Bezpečný prístup: metóda `.get()` a operátor `in`

Ak si nie sme istí, či kľúč existuje, hranaté zátvorky môžu spôsobiť `KeyError`. Bezpečnejšie sú:

- `slovnik.get("kluc")` — vráti hodnotu, alebo `None`, ak kľúč neexistuje
- `slovnik.get("kluc", predvolena)` — vráti predvolenú hodnotu, ak kľúč neexistuje
- `"kluc" in slovnik` — vráti `True`/`False` podľa toho, či kľúč v slovníku je


```python
# .get() nevyvolá chybu, aj keď kľúč neexistuje
print(kniha.get("zaujimava_kniha"))           # None
print(kniha.get("zaujimava_kniha", "chýba"))  # chýba
print(kniha.get("nazov"))                     # Ilias

# operátor in overí, či kľúč v slovníku existuje
print("nazov" in kniha)              # True
print("zaujimava_kniha" in kniha)    # False
```

## 6. Zmena hodnôt v slovníku

Hodnotu pri existujúcom kľúči zmeníme rovnakým spôsobom ako pri pridávaní.

## 7. Mazanie dát zo slovníka

```python
# Pokus o prístup k zmazanému kľúču vyvolá chybu
print(kniha["zaujimava_kniha"])
```

## 8. Užitočné metódy a funkcie

Slovník ponúka niekoľko užitočných metód a funkcií:

- `len(slovnik)` — počet položiek v slovníku
- `slovnik.keys()` — všetky kľúče
- `slovnik.values()` — všetky hodnoty
- `slovnik.items()` — dvojice `(kľúč, hodnota)`
- `type(slovnik)` — dátový typ


```python
print("Počet položiek:", len(kniha))
print("Kľúče:", kniha.keys())
print("Hodnoty:", kniha.values())
print("Dvojice:", kniha.items())
print("Typ:", type(kniha))
```

## 9. Vnorené štruktúry

Slovník môže obsahovať zoznamy aj ďalšie slovníky ako hodnoty.


```python
skola = {
    "nazov": "SPŠ IT",
    "mesto": "Tvrdošín",
    "triedy": ["I.C", "II.C", "III.C"],
    "riaditel": {
        "meno": "Ing. Novák",
        "email": "novak@skola.sk"
    }
}

# Prístup k vnoreným dátam
print(skola["triedy"][0])           # I.C
print(skola["riaditel"]["email"])    # novak@skola.sk
```
